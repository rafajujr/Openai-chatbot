### Text Toll Speak TTS Converte text em áudio


```python
import openai

client = openai.Client()
```

```python
texto = """
olá tudo bem!
Hoje vou te falar algo de extrema importância sobre a nossa vida. Para onde vamos após a morte?
a Palavra de Deus nos mostra em Hebreus 9.27 que "Após a morte, segue-se o juízo...". 
A destinação de nossa alma, após a morte é um dilema que nos permeiam ao longo da história da humanidade.
Segundo a bíblia, temos dois destinos eternos: ou estaremos juntos com Cristo nos céus, ou afastado dEle no inferno.
As nossas escolhas em vida são o que determinam para onde vamos. A principal delas é Crêr e aceitar Jesus como Senhor
e Salvador da nossa vida. Ele pagou o preço do pecado com a sua morte na cruz para que todos aqueles que confiam e crerem
 nEle, sejam salvos e sejam conduzidos ao céu, para viver um eternidade em paz com Deus, e com Cristo, o Senhor.
 Você já tomou essa decisão em sua vida?
 Decida hoje entregar sua vida para o Senhor Jesus e viver uma vida eterna em paz com Deus e com Cristo, o Senhor.
"""
```

```python
resposta = client.audio.speech.create(
    model='tts-1',
    voice='alloy',
    input=texto
)
```

```python
arquivo = 'files/audio.mp3'
resposta.write_to_file(arquivo)
```

### STT Converter áudio em texto

```python
audio = open('files/audio.mp3', 'rb')
transcricao = client.audio.transcriptions.create(
    model='whisper-1',
    file=audio
)
```

```python
print(transcricao.text)
```

### Trascrição como legenda

```python
audio = open('files/audio.mp3', 'rb')
transcript = client.audio.transcriptions.create(
  model="whisper-1", 
  file=audio,
  response_format="vtt", # Define o formato de saída
  language="pt",
  prompt="Legenda precisa em português brasileiro, usando pontuação correta e seguindo o ritmo da fala."
)
```

```python
print(transcript)
```
