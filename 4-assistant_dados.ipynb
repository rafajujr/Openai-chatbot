```python
import openai
import pandas as pd
```

```python
client = openai.Client()
```

```python
file = client.files.create(
    file=open("sales_data.csv", "rb"),
    purpose="assistants"
)
```

```python
print(file.id)
```

```python
assistant = client.beta.assistants.create(
    name="Analista de Dados",
    instructions="Você é um analista de dados. Use o code_interpreter para analisar o arquivo CSV fornecido e responder perguntas sobre previsões de receita.",
    tools=[{"type": "code_interpreter"}],
    tool_resources={
        "code_interpreter": {
            "file_ids": [file.id]
        }
    },
    model="gpt-4o"
)
```

```python
#pergunta = "Qual o maior produto vendido?"
pergunta = "gere um relatório de vendas por linha de produto"
```

```python
#Criação da Thread
thread = client.beta.threads.create()
message = client.beta.threads.messages.create(
    thread_id=thread.id,
    role="user",
    content=pergunta
)
```

```python
# Executa a thread
run = client.beta.threads.runs.create(
    thread_id=thread.id,
    assistant_id=assistant.id,
    instructions="Use o code_interpreter para analisar o arquivo CSV fornecido e responder perguntas sobre previsões de receita."
)
```

```python
# Aguarda a thread ser concluida
import time
while run.status in ["queued", "in_progress", "canceling"]:
    time.sleep(1)
    run = client.beta.threads.runs.retrieve(
        thread_id=thread.id,
        run_id=run.id
    )
    
```

```python
# Verificar resposta
if run.status == "completed":
    messagens = client.beta.threads.messages.list(
        thread_id=thread.id
    )
    print(messagens)
else:
    print(f"Erro: {run.status}")
```
