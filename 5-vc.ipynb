```python
import openai
from PIL import Image
import requests
```

```python
client = openai.Client()
```

### Gerar Imagens


```python
nome="robo_praia"
model="dall-e-3"
prompt="Crie uma imagem de dois robôs conversando sobre a biblia em uma praia"
size="1024x1024"
quality="hd"
style="vivid"
```

```python
resposta = client.images.generate(
    prompt=prompt,
    model=model,
    size=size,
    quality=quality,
    style=style,
    n=1
)
```

```python
resposta.data[0].revised_prompt
```

```python
nome_arquivo = f"files/{nome}_{model}_{quality}_{style}.png"
image_url = resposta.data[0].url
image_data = requests.get(image_url).content
with open(nome_arquivo, "wb") as f:
    f.write(image_data)
```

### Extraia placa de carro

```python
texto = 'Qual é a placa? devolva apenas a placa!'
```

```python
resposta = client.chat.completions.create(
    model='gpt-4.1-mini',
    messages=[{
        'role': 'user',
        'content': [
            {'type': 'text', 'text': texto},
            {'type': 'image_url', 'image_url': {
                'url': 'https://miltonjung.com.br/wp-content/uploads/2009/02/placabrasil.jpg'
            }}
        ]
    }],
    max_tokens=1000
)
```

```python
resposta.choices[0].message.content
```
