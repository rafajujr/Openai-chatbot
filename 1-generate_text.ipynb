### Utilizando função


```python
def geracao_texto(mensagem, modelo="gpt-3.5-turbo-0125", max_tokens=1000, temperature=0):
    resposta =  client.chat.completions.create(
        messages=mensagem,
        model=modelo,
        max_tokens=max_tokens,
        temperature=temperature
    )
    mensagem.append({"role": "assistant", "content": resposta.choices[0].message.content})
    return mensagem
```

```python
msg1 = [{"role": "user", "content": "Defina a diferenca entre tricotomia e dicotomia em poucas palavras, bem resumido"}]
mensagem = geracao_texto(msg1)
```

```python
mensagem
```

### Utlizando Stream

```python
mensagens = [{"role": "user", "content": "Conte uma história simples e objetiva sobre Jesus"}]
resposta = client.chat.completions.create(
    messages=mensagens,
    model="gpt-3.5-turbo-0125",
    max_tokens=1000,
    temperature=0,
    stream=True
)
```

```python
resposta
```

```python
for resposta_stream in resposta:
    texto = resposta_stream.choices[0].delta.content
    if texto:
        print(texto, end='')
```
